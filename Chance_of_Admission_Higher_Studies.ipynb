# Chance of Admission for Higher Studies

## Data Analysis and Machine Learning Project

## Project Objective

The objective of this project is to analyze different factors that may be associated with a student's chance of admission to higher studies.

The project uses student information such as GRE Score, TOEFL Score, University Rating, Statement of Purpose, Letter of Recommendation, Undergraduate CGPA, and Research Experience.

Machine learning techniques will be used to estimate the Chance of Admit based on these features.

## Dataset Features

The dataset contains the following features:

1. GRE Score - GRE examination score
2. TOEFL Score - TOEFL examination score
3. University Rating - Rating of the university
4. SOP - Statement of Purpose strength
5. LOR - Letter of Recommendation strength
6. CGPA - Undergraduate CGPA
7. Research - Research experience
8. Chance of Admit - Estimated chance of admission

## Feature Ranges

- GRE Score: 290 to 340
- TOEFL Score: 92 to 120
- University Rating: 1 to 5
- SOP: 1 to 5
- LOR: 1 to 5
- Undergraduate CGPA: 6.8 to 9.92
- Research Experience: 0 or 1
- Chance of Admit: approximately 0.34 to 0.97

```python
plt.figure(figsize=(8,5))

plt.hist(df["gre_score"], bins=10)

plt.xlabel("GRE Score")
plt.ylabel("Number of Students")
plt.title("Distribution of GRE Scores")

plt.show()
```

```python
plt.figure(figsize=(8,4))

plt.boxplot(df["gre_score"])

plt.ylabel("GRE Score")
plt.title("GRE Score Boxplot")

plt.show()
```

```python
correlation = df["gre_score"].corr(df["chance_of_admit"])

print("Correlation between GRE Score and Chance of Admit:", correlation)
```
