# Chance of Admission for Higher Studies

## Data Analysis and Machine Learning Project

## Project Objective

The objective of this project is to analyze different factors that may be associated with a student's chance of admission to higher studies.

The project uses student information such as GRE Score, TOEFL Score, University Rating, Statement of Purpose, Letter of Recommendation, Undergraduate CGPA, and Research Experience.

Machine learning techniques will be used to estimate the Chance of Admit based on these features.

## Dataset Features

The dataset contains the following features:

1. GRE Score - GRE examination score
2. TOEFL Score - TOEFL examination score
3. University Rating - Rating of the university
4. SOP - Statement of Purpose strength
5. LOR - Letter of Recommendation strength
6. CGPA - Undergraduate CGPA
7. Research - Research experience
8. Chance of Admit - Estimated chance of admission

## Feature Ranges

- GRE Score: 290 to 340
- TOEFL Score: 92 to 120
- University Rating: 1 to 5
- SOP: 1 to 5
- LOR: 1 to 5
- Undergraduate CGPA: 6.8 to 9.92
- Research Experience: 0 or 1
- Chance of Admit: approximately 0.34 to 0.97

```python

print("Checking dataset...")
print("Dataset shape:", df.shape)
```

```python

print("Dataset columns:")
print(df.columns.tolist())
```

```python

print("First 10 Research Values:")
print(df["research"].head(10))
```

```python

print("Research Experience Statistics")
print("--------------------------------")

print("Mean:", df["research"].mean())
print("Median:", df["research"].median())
print("Minimum:", df["research"].min())
print("Maximum:", df["research"].max())
print("Standard Deviation:", df["research"].std())
```

```python

print("Students without research experience:",
      (df["research"] == 0).sum())

print("Students with research experience:",
      (df["research"] == 1).sum())
```

```python

research_percentages = (
    df["research"].value_counts(normalize=True).sort_index() * 100
)

print("Research Experience Percentages:")
print(research_percentages.round(2))
```

```python

import matplotlib.pyplot as plt

research_counts = df["research"].value_counts().sort_index()

plt.figure(figsize=(7, 5))
plt.bar(research_counts.index, research_counts.values)

plt.title("Research Experience Distribution")
plt.xlabel("Research Experience (0 = No, 1 = Yes)")
plt.ylabel("Number of Students")
plt.xticks([0, 1])

plt.show()
```

```python

plt.figure(figsize=(6, 6))

plt.pie(
    research_counts.values,
    labels=["No Research", "Research Experience"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Percentage of Students by Research Experience")
plt.show()
```

```python

average_chance = df.groupby("research")[
    "chance_of_admit"
].mean()

print("Average Chance of Admit by Research Experience:")
print(average_chance.round(4))
```

```python

plt.figure(figsize=(7, 5))

plt.bar(
    average_chance.index,
    average_chance.values
)

plt.title("Research Experience vs Average Chance of Admit")
plt.xlabel("Research Experience (0 = No, 1 = Yes)")
plt.ylabel("Average Chance of Admit")
plt.xticks([0, 1])

plt.show()
```

```python

plt.figure(figsize=(7, 5))

df.boxplot(
    column="chance_of_admit",
    by="research"
)

plt.title("Chance of Admit by Research Experience")
plt.suptitle("")
plt.xlabel("Research Experience (0 = No, 1 = Yes)")
plt.ylabel("Chance of Admit")

plt.show()
```

```python

correlation = df["research"].corr(
    df["chance_of_admit"]
)

print("Correlation between Research Experience and Chance of Admit:")
print(round(correlation, 4))
```
